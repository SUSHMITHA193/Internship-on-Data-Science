```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

print("All libraries imported successfully!")
```

```python

# Load the uploaded CSV file
df_raw = pd.read_csv("weather.csv")

# Keep an untouched copy of the original data
df = df_raw.copy()

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())
```

```python

print("Dataset dimensions:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types and non-null counts:")
df.info()

print("\nFirst five records:")
display(df.head())

print("\nSummary statistics:")
display(df.describe(include="all").T)

print("\nMissing values per column:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate rows:", df.duplicated().sum())
```

```python

missing_counts = df.isnull().sum()
missing_counts = missing_counts[missing_counts > 0].sort_values()

if len(missing_counts) > 0:
    missing_counts.plot(kind="barh", figsize=(9, 5))
    plt.title("Missing Values by Column")
    plt.xlabel("Number of Missing Values")
    plt.ylabel("Column")
    plt.tight_layout()
    plt.savefig("visualization_1_missing_values.png", dpi=300)
    plt.show()
else:
    print("No missing values found.")
```

```python

# 1. Standardize column names
df.columns = (
    df.columns.str.strip()
    .str.replace(" ", "_")
)

# 2. Identify numeric and categorical columns
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

# 3. Convert numeric columns safely
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# 4. Remove duplicate rows
duplicates_before = df.duplicated().sum()
df = df.drop_duplicates().copy()

# 5. Record missing values before imputation
missing_before = df.isnull().sum()

# 6. Fill missing numerical values using the median
for col in numeric_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

# 7. Fill missing categorical values using the mode
for col in categorical_cols:
    if df[col].isnull().any():
        mode_values = df[col].mode()
        fill_value = mode_values.iloc[0] if not mode_values.empty else "Unknown"
        df[col] = df[col].fillna(fill_value)

print("Duplicates removed:", duplicates_before)
print("\nMissing values before cleaning:")
print(missing_before[missing_before > 0])

print("\nMissing values after cleaning:")
print(df.isnull().sum().sum())

print("\nCleaned dataset dimensions:", df.shape)
display(df.head())
```

```python

plt.figure(figsize=(9, 5))

sns.histplot(data=df, x="MaxTemp", bins=20, kde=True)

plt.title("Distribution of Maximum Temperature")
plt.xlabel("Maximum Temperature")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("visualization_2_temperature_distribution.png", dpi=300)
plt.show()
```

```python

plt.figure(figsize=(9, 5))

sns.histplot(data=df, x="Rainfall", bins=25, kde=True)

plt.title("Distribution of Daily Rainfall")
plt.xlabel("Rainfall")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("visualization_3_rainfall_distribution.png", dpi=300)
plt.show()
```

```python

correlation_cols = [
    "MinTemp", "MaxTemp", "Rainfall",
    "Evaporation", "Sunshine", "WindGustSpeed",
    "Humidity9am", "Humidity3pm",
    "Pressure9am", "Pressure3pm",
    "Cloud9am", "Cloud3pm",
    "Temp9am", "Temp3pm"
]

correlation_matrix = df[correlation_cols].corr()

plt.figure(figsize=(13, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between Weather Variables")
plt.tight_layout()
plt.savefig("visualization_4_correlation_heatmap.png", dpi=300)
plt.show()
```

```python

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="RainTomorrow")

plt.title("Distribution of Rain Tomorrow")
plt.xlabel("Rain Tomorrow")
plt.ylabel("Number of Observations")
plt.tight_layout()
plt.savefig("visualization_5_rain_tomorrow.png", dpi=300)
plt.show()

print(df["RainTomorrow"].value_counts())
print("\nPercentage distribution:")
print(df["RainTomorrow"].value_counts(normalize=True).mul(100).round(2))
```

```python

# Descriptive statistics
print("Weather variable summary:")
display(df.describe().T.round(2))

# Average weather measurements
print("\nAverage measurements:")
print("Mean maximum temperature:", round(df["MaxTemp"].mean(), 2))
print("Mean minimum temperature:", round(df["MinTemp"].mean(), 2))
print("Mean rainfall:", round(df["Rainfall"].mean(), 2))
print("Mean sunshine:", round(df["Sunshine"].mean(), 2))

# Highest recorded values
print("\nMaximum recorded rainfall:", df["Rainfall"].max())
print("Highest maximum temperature:", df["MaxTemp"].max())
print("Highest wind gust speed:", df["WindGustSpeed"].max())

# Strongest correlation pairs, excluding self-correlations
corr = df[correlation_cols].corr()
upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
strong_pairs = upper.stack().sort_values(
    key=lambda x: x.abs(), ascending=False
)

print("\nTop five absolute correlations:")
display(strong_pairs.head(5).to_frame("Correlation"))
```
